```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from patsy import dmatrices

from statsmodels.genmod.families import NegativeBinomial
from statsmodels.formula.api import glm

# Load your data
df = pd.read_excel('../data/sz_log/p1_log_combined.xlsx')

# Ensure proper types
df["Date"] = pd.to_datetime(df["Date"])
df["ASO_Dose"] = pd.to_numeric(df["ASO_Dose"], errors="coerce")
df["total_seizures"] = pd.to_numeric(df["total_seizures"], errors="coerce")

# Drop rows with missing values
df = df.dropna(subset=["ASO_Dose", "total_seizures"])


#Negative Binomial GLM
print('Baseline days')
print(sum(df['Post_ASO']==0))
print('Post_ASO days')
print(sum(df['Post_ASO']==1))

df["time"] = np.arange(len(df))
# Add lagged seizures
df["lag1"] = df["total_seizures"].shift(1)

# Fit Negative Binomial GLM
model = glm("total_seizures ~ time + Post_ASO + ASO_Dose + lag1",
            data=df,family=sm.families.NegativeBinomial(alpha=0.5)).fit()

# Print model summary
print(model.summary())
# Goodness-of-fit metrics
# Deviance and Pearson Chi-Square
print(f"Deviance: {model.deviance:.2f}")
print(f"Degrees of Freedom (Residual): {model.df_resid}")
print(f"Pearson Chi-Square: {model.pearson_chi2:.2f}")

# Dispersion ratio 
dispersion_ratio = model.pearson_chi2 / model.df_resid
print(f"Dispersion Ratio: {dispersion_ratio:.2f}")
```
